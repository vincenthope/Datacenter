```python
# Simple AI Movie Recommendation System

print("Welcome to Movie Recommender")

genre = input("What genre do you like? (action/comedy/romance): ")

if genre == "action":
    print("Recommended Movies:")
    print("1. Avengers")
    print("2. Batman")

elif genre == "comedy":
    print("Recommended Movies:")
    print("1. Mr Bean")
    print("2. The Mask")

elif genre == "romance":
    print("Recommended Movies:")
    print("1. Titanic")
    print("2. The Notebook")

else:
    print("Sorry, no recommendations found")
```

```python
movies = {
    "action": ["Avengers", "Batman"],
    "comedy": ["Mr Bean", "The Mask"],
    "romance": ["Titanic", "The Notebook"]
}

genre = input("Enter genre: ")

if genre in movies:
    print("Recommended movies:")
    
    for movie in movies[genre]:
        print(movie)

else:
    print("Genre not found")
```
